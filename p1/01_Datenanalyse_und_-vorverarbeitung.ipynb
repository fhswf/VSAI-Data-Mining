### Struktur des Datensatzes
- **Anzahl der Instanzen (Beispiele)**: 150
- **Anzahl der Attribute (Merkmale)**: 5
  - **SepalLength**: Länge des Sepals in Zentimetern
  - **SepalWidth**: Breite des Sepals in Zentimetern
  - **PetalLength**: Länge der Blüte in Zentimetern
  - **PetalWidth**: Breite der Blüte in Zentimetern
  - **Species**: Art der Iris-Pflanze (3 Klassen):
    - Iris-setosa
    - Iris-versicolor
    - Iris-virginica

Wir laden den Datensatz im Folgenden mit der Funktion `pd.read_csv()`.

```python
# Beispiel: Iris-Datensatz
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
column_names = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth', 'Species']
data = pd.read_csv(url, names=column_names)
data.head(10)
```

`data.info()` gibt eine Übersicht über den Pandas DataFrame des Iris-Datensatzes aus. Dabei werden die Anzahl der Einträge und Spalten, die Spaltennamen, Datentypen, die Anzahl der Nicht-Null-Werte sowie der verwendete Speicher angezeigt.

```python
# Informationen über das Dataset
data.info()
```

`data.describe()` erstellt eine statistische Zusammenfassung der numerischen Spalten des Iris-Datensatzes. Angezeigt werden unter anderem Anzahl, Mittelwert, Standardabweichung, Minimum, Quartile und Maximum der jeweiligen Merkmale.

```python
# Statistische Zusammenfassung
data.describe()
```

```python
# Fehlende Werte überprüfen
data.isnull().sum()
```

Der Befehl [`sns.pairplot(data, hue='Species')`](https://seaborn.pydata.org/generated/seaborn.pairplot.html) aus der [Seaborn-Bibliothek](https://seaborn.pydata.org/) im folgenden Codeabschnitt ist ähnlich der [Pandas](https://pandas.pydata.org/) Funktion [`pd.plotting.scatter_matrix(data)`](https://pandas.pydata.org/docs/reference/api/pandas.plotting.scatter_matrix.html). Er generiert eine komplette Übersicht für alle numerischen Merkmale im Datensatz:

- Für jede Kombination von Spalten wird ein Scatterplot erstellt.
- Beispiel mit Iris: SepalLength vs SepalWidth, SepalLength vs PetalLength, usw.
- Auf der Diagonale werden meist Verteilungsplots (Histogramm/KDE) für jede Variable angezeigt.
- `hue='Species'` färbt die Punkte je nach Art ein, z. B. setosa, versicolor, virginica.

Das ist besonders nützlich, um:
- Beziehungen zwischen mehreren Merkmalen zu entdecken
- Muster oder Cluster zwischen Klassen zu sehen
- Ausreißer oder Überschneidungen zu erkennen

Vergleich `pairplot()` zu `scatter_matrix()`:
- **`sns.pairplot(data, hue='Species')`** gehört zu [Seaborn](https://seaborn.pydata.org/). Mit `hue` lassen sich Punkte nach einer Kategorie einfärben - hier nach `Species`. Das erleichtert den Vergleich der Iris-Arten.
- **`pd.plotting.scatter_matrix(data)`** gehört zu [Pandas](https://pandas.pydata.org/) und bietet eine einfachere, direkt in Pandas integrierte Darstellung. Eine Gruppierung und Einfärbung nach `Species` ist nicht so direkt vorgesehen; dafür müssen die Gruppen meist separat geplottet werden.
- Seaborn bietet mehr Möglichkeiten zur Gestaltung und Anpassung, während `scatter_matrix()` für einen schnellen Überblick praktisch ist.

```python
sns.pairplot(data, hue='Species')
plt.show()
```

Im Pairplot sind die Beziehungen zwischen den vier numerischen Merkmalen des Iris-Datensatzes dargestellt. Die Farben kennzeichnen die drei Iris-Arten. Besonders anhand von PetalLength und PetalWidth lassen sich die Arten gut unterscheiden: Iris-setosa bildet eine klar getrennte Gruppe, während sich Iris-versicolor und Iris-virginica teilweise überschneiden. Auf der Diagonale sind die Verteilungen der einzelnen Merkmale zu sehen.
